# Air Quality Analysis: Finding the Best Neighborhood in Pittsburgh

## Introduction
For the final project, I chose to use air quality as my dataset. My approach was to take the data and assign every different level of air quality a number value, "Good": 2, "Moderate": 1, "Unhealthy": -1, "Very unhealthy": -2, "Hazardous": -3, "Bad": -2. Then I wrote a program to run through the whole dataset and apply these values to each individual neighborhood, this was then divided by how often each area appears in the list. The three areas that are the best on air quality are ranked by the output of the program.


```python
import csv
from collections import defaultdict

file_path = "air_quality_data.csv"

# Scaled point system for all descriptions
description_points = {
    "Good": 2,
    "Moderate": 1,
    "Unhealthy": -1,
    "Very unhealthy": -2,
    "Hazardous": -3,
    "Bad": -2
}

def update_site_scores(file_path):
    scores = defaultdict(int)
    counts = defaultdict(int)

    with open(file_path, newline='') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            site = row["site"].strip()
            description = row["description"].strip().capitalize()

            points = description_points.get(description, 0)
            scores[site] += points
            counts[site] += 1  # Track how many times each site appears

    # Compute average score per site
    averages = {site: scores[site] / counts[site] for site in scores}

    return averages

site_averages = update_site_scores(file_path)

# Sort by average score (descending)
sorted_sites = sorted(site_averages.items(), key=lambda x: x[1], reverse=True)

print("Areas ranked by air quality (best to worst):")
for rank, (site, avg_score) in enumerate(sorted_sites, start=1):
    print(f"{rank}. {site}: {avg_score:.2f} average points")
```
