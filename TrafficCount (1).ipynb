## City of Pittsburgh Traffic Data Count
i will be using Pittsburgh traffic data count by neighborhoods to find the safest drivers. For that i will calculate the max speed, and median speed.

https://data.wprdc.org/datastore/dump/6dfd4f8f-cbf5-4917-a5eb-fd07f4403167

Improted Data from the table and looked for the neighborhoods and speed limits in them, to find if people were speeding or not, if it didnt have any data on the csv file i didn't use that data

```python
import pandas as pd

#loaded the data using read_csv
df = pd.read_csv('TrafficCount.csv')

#kept only the needed columns, and removed the empty ones.
cols_needed = ['neighborhood', 'speed_limit', 'median_speed']
speed_df = df[cols_needed].dropna(subset=cols_needed)

# converting columns into numbers
speed_df['speed_limit'] = pd.to_numeric(speed_df['speed_limit'], errors='coerce')
speed_df['median_speed'] = pd.to_numeric(speed_df['median_speed'], errors='coerce')

# dropped NaNs after the conversion
speed_df = speed_df.dropna(subset=['speed_limit', 'median_speed'])

# grouped the neighborhood and calculated the median speed
by_neigh = speed_df.groupby('neighborhood').agg({'speed_limit':'median', 'median_speed': 'median'}).reset_index()

# found the best neighborhood where the median speed was less than the actual speed limit
by_neigh['difference'] = by_neigh['speed_limit'] - by_neigh['median_speed']
best = by_neigh[by_neigh['median_speed'] <= by_neigh['speed_limit']].sort_values('difference', ascending=False).head(10)

# found worst neighborhoods where median speed was above the actual speed limit
worst = by_neigh[by_neigh['median_speed'] > by_neigh['speed_limit']].sort_values('difference').head(10)
# printing out the top 10 best neighborhoods
print("Top 10 Best Neighborhoods:")
print(best[['neighborhood', 'speed_limit', 'median_speed', 'difference']])
#printing out the top 10 worst neighborhoods
print("Top 10 Worst Neighborhoods:")
print(worst[['neighborhood', 'speed_limit', 'median_speed', 'difference']])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 6))

# top 10 best neighborhoods
plt.subplot(1, 2, 1)
plt.bar(best['neighborhood'], best['difference'], color='green')
plt.title('Top 10 Best Neighborhoods')
plt.xlabel('Neighborhood')
plt.ylabel('Speed limit(mph)')
plt.xticks(rotation=90)


# top 10 worst neighborhoods
plt.subplot(1, 2, 2)
plt.bar(worst['neighborhood'], worst['difference'], color='red')
plt.title('Top 10 Worst Neighborhoods')
plt.xlabel('Neighborhood')
plt.ylabel('Speed limit(mph)')
plt.xticks(rotation=90)

plt.tight_layout()

# display the chart
plt.show()
```
